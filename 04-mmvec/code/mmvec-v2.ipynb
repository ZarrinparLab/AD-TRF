### AD mmvec

```python
#mtb
biom convert \
 -i mmvec_mtb_shared_feattab.txt \
 -o mmvec_mtb_shared_feattab.biom \
 --table-type="OTU table" \
 --to-hdf5

qiime tools import \
 --input-path mmvec_mtb_shared_feattab.biom \
 --output-path mmvec_mtb_shared_feattab.qza \
 --type FeatureTable[Frequency]

#mtg
biom convert \
 -i mmvec_mtg_shared_feattab.txt \
 -o mmvec_mtg_shared_feattab.biom \
 --table-type="OTU table" \
 --to-hdf5

qiime tools import \
 --input-path mmvec_mtg_shared_feattab.biom \
 --output-path mmvec_mtg_shared_feattab.qza \
 --type FeatureTable[Frequency]
```

```python
qiime mmvec paired-omics \
        --i-microbes mmvec_mtg_shared_feattab.qza \
        --i-metabolites mmvec_mtb_shared_feattab.qza \
        --p-summary-interval 1 \
        --m-metadata-file mmvec_comb_metadata.txt \
        --p-training-column Testing \
        --output-dir AD_summary_shared

qiime mmvec heatmap \
  --i-ranks AD_summary_shared/conditionals.qza \
  --m-microbe-metadata-file mmvec_mtg_feat_annotation.txt \
  --m-microbe-metadata-column Genus \
  --m-metabolite-metadata-file mmvec_mtb_feat_annotation.txt \
  --m-metabolite-metadata-column subclass \
  --o-visualization AD_summary_shared/ranks-heatmap.qzv
```

```python
qiime mmvec summarize-single \
        --i-model-stats model_summary/model_stats.qza \
        --o-visualization model-summary.qzv

# Null model with only biases
qiime mmvec paired-omics \
        --i-microbes otus_nt.qza \
        --i-metabolites lcms_nt.qza \
        --p-latent-dim 0 \
        --p-summary-interval 1 \
        --output-dir null_summary

qiime mmvec summarize-paired \
        --i-model-stats model_summary/model_stats.qza \
        --i-baseline-stats null_summary/model_stats.qza \
        --o-visualization paired-summary.qzv
```

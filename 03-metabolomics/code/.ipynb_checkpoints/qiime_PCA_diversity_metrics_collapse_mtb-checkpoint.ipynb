## **Calculating the alpha diversity**

We used the qiime plugin to calculate shannon diversity, observed OTUs, and Faith's PD. The latter we generated using the phylogenetic tree 'qemistree_serum.qza' generated through qemistree.

```python
#The alpha diveristy using 'shannon' 
! qiime diversity alpha \
--i-table /mnt/zarrinpar/scratch/sfloresr/AD_EM/data_files/AD_GNPS_BA/nf_output/clustering/gnps_quant_collapseiin_sum_corrgroup_feces.qza \
--p-metric shannon \
--o-alpha-diversity /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/alpha_shannon_lcms.qza
```

```python
! qiime tools export \
--input-path /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/alpha_shannon_lcms.qza \
--output-path /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/alpha_shannon_lcms
```

```python
#The alpha diveristy using 'observed_feature'  
! qiime diversity alpha \
--i-table /mnt/zarrinpar/scratch/sfloresr/AD_EM/data_files/AD_GNPS_BA/nf_output/clustering/gnps_quant_collapseiin_sum_corrgroup_feces.qza \
--p-metric observed_features \
--o-alpha-diversity /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/alpha_obsOTU_lcms.qza
```

```python
! qiime tools export \
--input-path /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/alpha_obsOTU_lcms.qza \
--output-path /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/alpha_obsOTU_lcms
```

## **Calculating the beta diversity**
To do this we used the qiime plugin to calculate bray curtis, aitchison, weighted and unweighted unifrac. The latter two were generated using the phylogenetic tree qemistree_serum.qza generated through qemistree.

```python
#The beta diversity using 'bray curtis' #rerun using 2024 
! qiime diversity beta \
--i-table /mnt/zarrinpar/scratch/sfloresr/AD_EM/data_files/AD_GNPS_BA/nf_output/clustering/gnps_quant_collapseiin_sum_corrgroup_feces.qza \
--p-metric braycurtis \
--o-distance-matrix /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/beta_braycurtis_lcms.qza
```

```python
! qiime tools export \
--input-path /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/beta_braycurtis_lcms.qza \
--output-path /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/beta_braycurtis_lcms
```

```python
#The beta diversity using 'aitchison'  
! qiime diversity beta \
--i-table /mnt/zarrinpar/scratch/sfloresr/AD_EM/data_files/AD_GNPS_BA/nf_output/clustering/gnps_quant_collapseiin_sum_corrgroup_feces.qza \
--p-metric aitchison \
--o-distance-matrix /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/beta_aitchison_lcms.qza
```

```python
! qiime tools export \
--input-path /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/beta_aitchison_lcms.qza \
--output-path /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/beta_aitchison_lcms
```

### Deicode (to calculate aitchison)-Fecal

In additon to calculating these metric through the default plugin methods, I also wanted to use this new one that uses log abundances to account for sparse mtbs as the previous methods do not correct for this (very possible) bias.

```python
! qiime deicode rpca \
    --i-table /mnt/zarrinpar/scratch/sfloresr/AD_EM/data_files/AD_GNPS_BA/nf_output/clustering/gnps_quant_collapseiin_sum_corrgroup_feces_forrpca.qza \
    --p-min-feature-count 10 \
    --p-min-sample-count 500 \
    --o-biplot /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/beta_deicode_ordination.qza \
    --o-distance-matrix /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/beta_deicode_lcms.qza
```

```python
! qiime emperor biplot \
    --i-biplot /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/beta_deicode_ordination.qza \
    --m-sample-metadata-file /mnt/zarrinpar/scratch/sfloresr/AD_EM/data_files/FBMN/metadata_table/metadata_table-noblanks.txt \
    --p-ignore-missing-samples \
    --o-visualization /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/beta_deicode_biplot.qzv
```

```python
qiime diversity beta-group-significance \
    --i-distance-matrix /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/beta_deicode_lcms.qza  \
    --m-metadata-file /mnt/zarrinpar/scratch/sfloresr/AD_EM/data_files/FBMN/metadata_table/metadata_table-noblanks.txt \
    --m-metadata-column ATTRIBUTE_sex \
    --p-method permanova \
    --p-pairwise \
    --o-visualization /mnt/zarrinpar/scratch/sfloresr/AD_EM/diversity_metrics/fecal/beta_deicode_sex-significance.qzv
```
